```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
```

```python
image_folder = r"C:\Users\devah\Documents\ICT MU\Sem 5\DSIP\Lab\OPE9 images"
```

```python
image_paths = [
    glob.glob(os.path.join(image_folder, "image1_noisy_face.*"))[0],
    glob.glob(os.path.join(image_folder, "image2_noisy_object.*"))[0],
    glob.glob(os.path.join(image_folder, "image3_city.*"))[0],
    glob.glob(os.path.join(image_folder, "image4_blurred_building.*"))[0],
    glob.glob(os.path.join(image_folder, "image5_motion_blur.*"))[0]
]

for path in image_paths:
    print(path)
```

```python
images = []

for path in image_paths:
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

    if img is None:
        print("Image not found:", path)
    else:
        images.append(img)
        print("Loaded:", os.path.basename(path))

print("\nNumber of images loaded:", len(images))
```

```python
image_1 = images[0]
image_2 = images[1]
image_3 = images[2]
image_4 = images[3]
image_5 = images[4]
```

```python
plt.figure(figsize=(15, 8))

for i, image in enumerate(images):
    plt.subplot(2, 3, i + 1)
    plt.imshow(image, cmap="gray")
    plt.title(f"Original Image {i + 1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
mean_image_1 = cv2.blur(image_1, (5, 5))

plt.figure(figsize=(6, 5))
plt.imshow(mean_image_1, cmap='gray')
plt.title("Image 1 - Mean Filter (5x5)")
plt.axis("off")
plt.show()
```

```python
gaussian_image_1 = cv2.GaussianBlur(
    image_1,
    (5, 5),
    0
)

plt.figure(figsize=(6, 5))
plt.imshow(gaussian_image_1, cmap='gray')
plt.title("Image 1 - Gaussian Filter (5x5)")
plt.axis("off")
plt.show()
```

```python
median_image_1 = cv2.medianBlur(
    image_1,
    5
)

plt.figure(figsize=(6, 5))
plt.imshow(median_image_1, cmap='gray')
plt.title("Image 1 - Median Filter (5x5)")
plt.axis("off")
plt.show()
```

```python
laplacian_1 = cv2.Laplacian(
    image_1,
    cv2.CV_64F
)

laplacian_1 = cv2.convertScaleAbs(laplacian_1)

sharpened_laplacian_1 = cv2.addWeighted(
    image_1,
    1.0,
    laplacian_1,
    1.0,
    0
)

plt.figure(figsize=(6, 5))
plt.imshow(sharpened_laplacian_1, cmap='gray')
plt.title("Image 1 - Laplacian Sharpening")
plt.axis("off")
plt.show()
```

```python
blurred_1 = cv2.GaussianBlur(
    image_1,
    (5, 5),
    0
)

unsharp_1 = cv2.addWeighted(
    image_1,
    1.5,
    blurred_1,
    -0.5,
    0
)

plt.figure(figsize=(6, 5))
plt.imshow(unsharp_1, cmap='gray')
plt.title("Image 1 - Unsharp Masking")
plt.axis("off")
plt.show()
```

```python
blurred_hb_1 = cv2.GaussianBlur(image_1,(5, 5),0)
high_boost_1 = cv2.addWeighted(image_1,2.0,blurred_hb_1,-1.0,0)
plt.figure(figsize=(6, 5))
plt.imshow(high_boost_1, cmap='gray')
plt.title("Image 1 - High-Boost Filtering")
plt.axis("off")
plt.show()
```
