```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# REPLACE THE PATH BELOW WITH THE PATH TO YOUR IMAGE ON YOUR COMPUTER
image_path = r"C:\Users\DileeP Kumar\OneDrive\Pictures\1294623.png"

src_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if src_image is None:
    print(f"Could not load image at {image_path}. Please check the filename.")
else:
    negative_image = 255 - src_image
    
    # Changed threshold from 128 to 100
    _, thresholded_image = cv2.threshold(src_image, 100, 255, cv2.THRESH_BINARY)
    
    # Changed gamma from 2.0 to 0.5
    gamma = 0.5
    normalized_image = src_image / 255.0
    gamma_corrected_image = np.power(normalized_image, 1/gamma)
    gamma_corrected_image = np.uint8(gamma_corrected_image * 255)
    
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 4, 1)
    plt.imshow(src_image, cmap='hot')
    plt.title('Original')
    plt.axis('off')
    
    plt.subplot(1, 4, 2)
    plt.imshow(negative_image, cmap='hot')
    plt.title('Negative')
    plt.axis('off')
    
    plt.subplot(1, 4, 3)
    plt.imshow(thresholded_image, cmap='hot')
    plt.title('Threshold')
    plt.axis('off')
    
    plt.subplot(1, 4, 4)
    plt.imshow(gamma_corrected_image, cmap='hot')
    plt.title('Gamma Corrected')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
```
