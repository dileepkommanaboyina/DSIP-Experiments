```python
import numpy as np
import matplotlib.pyplot as plt

N = 5
n = np.arange(-3, 15)
pattern = [4, 1, 3, 0, 2]
x = np.array([pattern[i % N] for i in n])

plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('Periodic Discrete-Time Signal')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2, 8)
x = np.zeros_like(n)

x[(n >= 0) & (n < 2)] = 2
x[(n >= 2) & (n < 5)] = 5
x[(n >= 5) & (n < 6)] = 3
x[(n >= 6) & (n < 7)] = 1

plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('Generated Staircase Signal')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 6)
x = np.zeros_like(n)

x[(n >= -4) & (n <= -2)] = -3
x[(n >= -1) & (n <= 1)] = n[(n >= -1) & (n <= 1)] + 1
x[(n >= 2) & (n <= 4)] = 4

plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('Piecewise Signal')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-4, 12)

def u(n_val):
    return np.where(n_val >= 0, 1, 0)

x = 2 * u(n + 1) - 2 * u(n - 2) - 4 * u(n - 6)

plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('x[n] = 2u[n+1] - 2u[n-2] - 4u[n-6]')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 6)

def delta(n_val):
    return np.where(n_val == 0, 1, 0)

x = 4 * delta(n + 2) - 2 * delta(n - 1) + 3 * delta(n - 3)

plt.figure(figsize=(8, 4))
plt.stem(n, x)
plt.title('x[n] = 4δ[n+2] - 2δ[n-1] + 3δ[n-3]')
plt.xlabel('n')
plt.ylabel('x[n]')
plt.grid(True)
plt.show()
```
