```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2, 4, 1, 5, 3])

plt.figure(figsize=(6, 4))
plt.stem(range(len(x)), x)
plt.title("Discrete-Time Signal x[n]")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

y = np.array([1, 3, 0, 4, 2])

plt.figure(figsize=(6, 4))
plt.stem(range(len(y)), y)
plt.title("Discrete-Time Signal y[n]")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2, 4, 1, 5, 3])

auto_corr = np.correlate(x, x, mode='full')
lags = np.arange(-(len(x)-1), len(x))

plt.figure(figsize=(6, 4))
plt.stem(lags, auto_corr)
plt.title("Autocorrelation of x[n]")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2, 4, 1, 5, 3])
y = np.array([1, 3, 0, 4, 2])

cross_corr = np.correlate(x, y, mode='full')
lags = np.arange(-(len(x)-1), len(x))

plt.figure(figsize=(6, 4))
plt.stem(lags, cross_corr)
plt.title("Cross-Correlation of x[n] and y[n]")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([2, 4, 1, 5, 3])
y = np.array([1, 3, 0, 4, 2])

auto_corr = np.correlate(x, x, mode='full')
cross_corr = np.correlate(x, y, mode='full')
lags = np.arange(-(len(x)-1), len(x))

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.stem(lags, auto_corr)
plt.title("Autocorrelation")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.stem(lags, cross_corr)
plt.title("Cross-Correlation")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
