```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = r"C:\Users\DileeP Kumar\OneDrive\Pictures\Black Clover Asta and Noelle Silva.jpg"
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise FileNotFoundError(f"Could not load image: {image_path}")

# Hit-or-miss expects a binary image with values 0 and 1.
_, binary_image = cv2.threshold(image, 127, 1, cv2.THRESH_BINARY)

pattern = np.array([
    [-1,  1, -1],
    [ 1,  1,  1],
    [-1,  1, -1]
], dtype=np.int8)

result = cv2.morphologyEx(binary_image, cv2.MORPH_HITMISS, pattern)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(image, cmap="gray")
axes[0].set_title("Original image")
axes[1].imshow(result, cmap="gray", vmin=0, vmax=1)
axes[1].set_title("Hit-or-miss result")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
